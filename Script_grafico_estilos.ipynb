```python
# ==========================================
# GRÁFICO DE DISTRIBUIÇÃO DE ESTILOS
# ==========================================

import matplotlib.pyplot as plt
import numpy as np

# Dados corrigidos com base na análise
modelos = ['BERTimbau', 'Qwen 1.8B']
categorias = ['Acadêmico', 'Narrativo', 'Descritivo']

# Quantidade de artigos por categoria e modelo
# Fonte: resultados da classificação textual
dados = {
    'BERTimbau': [3, 6, 21],      # Total: 30 artigos
    'Qwen 1.8B': [3, 4, 23]       # Total: 30 artigos
}

# Configuração do gráfico
x = np.arange(len(categorias))  # Posições das categorias
largura = 0.35  # Largura das barras

fig, ax = plt.subplots(figsize=(10, 6))

# Barras para BERTimbau
rects1 = ax.bar(x - largura/2, dados['BERTimbau'], largura,
                label='BERTimbau', color='#2E86C1', edgecolor='black', linewidth=1)

# Barras para Qwen 1.8B
rects2 = ax.bar(x + largura/2, dados['Qwen 1.8B'], largura,
                label='Qwen 1.8B', color='#E67E22', edgecolor='black', linewidth=1)

# Rótulos e títulos
ax.set_xlabel('Categorias de Estilo', fontsize=12, fontweight='bold')
ax.set_ylabel('Quantidade de Artigos', fontsize=12, fontweight='bold')
#ax.set_title('Distribuição de Estilos nos 30 Artigos', fontsize=14, fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels(categorias, fontsize=11)

ax.legend(loc='upper left', fontsize=11)

# Adicionar valores sobre as barras
def autolabel(rects):
    """Adiciona rótulos de valor acima de cada barra."""
    for rect in rects:
        height = rect.get_height()
        ax.annotate(f'{int(height)}',
                    xy=(rect.get_x() + rect.get_width() / 2, height),
                    xytext=(0, 3),  # 3 pontos de deslocamento vertical
                    textcoords="offset points",
                    ha='center', va='bottom', fontsize=10, fontweight='bold')

autolabel(rects1)
autolabel(rects2)

# Ajustar limites do eixo y para acomodar os rótulos
ax.set_ylim(0, 25)

# Grade para melhor legibilidade
ax.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()

# Salvar a figura (opcional)
# plt.savefig('distribuicao_estilos.png', dpi=300, bbox_inches='tight')
plt.savefig('distribuicao_estilos.pdf', dpi=300, bbox_inches='tight', format='pdf')

plt.show()

# ==========================================
# TABELA RESUMO
# ==========================================

print("\n" + "="*50)
print("RESUMO DA DISTRIBUIÇÃO DE ESTILOS")
print("="*50)
print(f"\n{'Categoria':<15} {'BERTimbau':<12} {'Qwen 1.8B':<12}")
print("-"*50)
for i, cat in enumerate(categorias):
    print(f"{cat:<15} {dados['BERTimbau'][i]:<12} {dados['Qwen 1.8B'][i]:<12}")
print("-"*50)
print(f"{'Total':<15} {sum(dados['BERTimbau']):<12} {sum(dados['Qwen 1.8B']):<12}")
```
